```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Load the dataset
df = pd.read_csv("student_performance.csv")
df.head()
```

## 1. Understanding the Dataset
First, we inspect the shape, columns, data types, and basic statistics.


```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nBasic statistics:")
display(df.describe(include="all"))
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

## 2. Handling Missing Values
Numeric score columns are filled with their median values. Median is useful because it is less affected by extreme values than the mean.


```python
numeric_columns = ["Age", "Python_Score", "Pandas_Score", "ML_Score"]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing values after imputation:")
print(df.isnull().sum())
```

## 3. Detecting and Removing Duplicate Records
Duplicate rows can cause incorrect analysis and model results, so they are removed.


```python
print("Duplicate rows before removal:", df.duplicated().sum())

df = df.drop_duplicates().reset_index(drop=True)

print("Duplicate rows after removal:", df.duplicated().sum())
print("New dataset shape:", df.shape)
```

## 4. Encoding Categorical Data
Machine learning algorithms generally work with numerical values. We encode the `Gender` and `Placed` columns using LabelEncoder.


```python
label_encoder = LabelEncoder()

df["Gender_Encoded"] = label_encoder.fit_transform(df["Gender"])
df["Placed_Encoded"] = label_encoder.fit_transform(df["Placed"])

df[["Gender", "Gender_Encoded", "Placed", "Placed_Encoded"]].head()
```

## 5. Feature Scaling
The score columns are standardized using `StandardScaler` so that they have a comparable scale.


```python
score_columns = ["Python_Score", "Pandas_Score", "ML_Score"]

scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[score_columns] = scaler.fit_transform(df_scaled[score_columns])

df_scaled[score_columns].head()
```

## 6. Simple Data Analysis
We calculate the average score for each subject and compare the number of placed and non-placed students.


```python
average_scores = df[score_columns].mean()
print("Average scores:")
print(average_scores.round(2))

print("\nPlacement distribution:")
print(df["Placed"].value_counts())
```

```python
# Visualize average scores
average_scores.plot(kind="bar", title="Average Score by Subject", xlabel="Subject", ylabel="Average Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Visualize placement distribution
df["Placed"].value_counts().plot(kind="bar", title="Placement Distribution", xlabel="Placement Status", ylabel="Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
